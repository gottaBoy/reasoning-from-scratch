# 本地实战 P1-03：多头 Attention 拆头与合头

P1-02 已通过。现在只练 view、transpose、reshape 和 stride，不加入 Linear、RoPE、GQA 或 Attention 分数。
先看 [P1-03教程](../docs/PRACTICE_P1_03_LAYOUT.md) 和中文模板，再在 `learning/my_ops.py` 实现 `split_heads` 与 `merge_heads`。


````python
import torch

B, T, H, D = 1, 2, 2, 3
x = torch.tensor([[[0., 1., 2., 3., 4., 5.],
                   [10., 11., 12., 13., 14., 15.]]])
split = x.view(B, T, H, D)
heads = split.transpose(1, 2)
print('x:', x.shape, x.stride())
print('split:', split.shape, split.stride())
print('heads:', heads.shape, heads.stride())
print(heads.tolist())
assert heads.tolist() == [[[[0., 1., 2.], [10., 11., 12.]],
                            [[3., 4., 5.], [13., 14., 15.]]]]
````

## 先预测

预测 `heads.transpose(1,2).reshape(B,T,H*D)` 和直接 `heads.reshape(B,T,H*D)` 的结果。
注意 `T=2`，这样漏掉 transpose 的错误不会被隐藏。


````python
correct = heads.transpose(1, 2).reshape(B, T, H * D)
wrong = heads.reshape(B, T, H * D)
print('correct:', correct.tolist())
print('wrong:', wrong.tolist())
assert torch.equal(correct, x)
assert not torch.equal(wrong, x)
````
