# 本地实战 P1-04：MHA / GQA / MQA

P1-03 已通过。现在只实现紧凑 KV head 到 Query head 的映射，不加入完整 Attention 或 Cache。
教程：[P1-04 GQA](../docs/PRACTICE_P1_04_GQA.md)。个人实现写在 `learning/my_ops.py` 的 `expand_kv_heads`。


````python
import torch

kv = torch.tensor([[[[0., 1.], [10., 11.]],
                   [[2., 3.], [12., 13.]]]])
expanded = kv.repeat_interleave(2, dim=1)
print('compact:', kv.shape)
print('expanded:', expanded.shape)
print(expanded.tolist())
assert expanded[:, 0].equal(kv[:, 0])
assert expanded[:, 1].equal(kv[:, 0])
assert expanded[:, 2].equal(kv[:, 1])
assert expanded[:, 3].equal(kv[:, 1])
````

## 先预测

对于 `H_q=4`：H_kv=4 是 MHA，H_kv=2 是 GQA，H_kv=1 是 MQA。
注意 Query 数量没有减少，减少的是独立 K/V head 数。
