把服务分成前台与车间，不是HTTP返回成功就代表所有内层优化正确：

```text
客户端 → API校验/鉴权 → 引擎/调度 → 模型加载与执行 → 返回输出
                         ↘ KV管理/实际Attention后端
```

探针发送temperature=0、max_tokens=8的小请求；
得到HTTP 200和文本，只证明这次协议/加载/推理链路可达。
它不证明分页kernel、请求取消、首token延迟或大规模容量。
